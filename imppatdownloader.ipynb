```python
import os
import re
import shutil
import subprocess
from pathlib import Path


ID_FILE = Path("ids.txt")


OUTPUT_DIR = Path(r"D:\SM")


DELAY = 2

# output folder

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"Output folder: {OUTPUT_DIR}")

# read

with open(ID_FILE, "r", encoding="utf-8") as f:
    text = f.read()

# Find IDs such as IMPHY000877 (test 2)
ids = re.findall(r"IMPHY\d+", text)

# Remove duplicates
ids = list(dict.fromkeys(ids))

print(f"Found {len(ids)} IMPHY IDs.\n")

# download

successful = []
failed = []

for count, imp_id in enumerate(ids, start=1):

    # IMPHY000877 -> 877
    numeric_id = int(imp_id.replace("IMPHY", ""))

    print("=" * 60)
    print(f"[{count}/{len(ids)}] Downloading {imp_id}")
    print("=" * 60)

    # Save each compound's downloader output in D:\SM
    env = os.environ.copy()
    env["PYTHONIOENCODING"] = "utf-8"
    env["PYTHONUTF8"] = "1"

    try:

        result = subprocess.run(
            [
                "imppat_downloader",
                "--start", str(numeric_id),
                "--end", str(numeric_id + 1),
                "--delay", str(DELAY),
                "--skip-existing"
            ],
            cwd=str(OUTPUT_DIR),
            env=env,
            text=True,
            encoding="utf-8",
            errors="replace",
            capture_output=True
        )

        
        if result.stdout:
            print(result.stdout)

        if result.stderr:
            print(result.stderr)

        # find

        sdf_files = list(OUTPUT_DIR.rglob("*.sdf"))

        # Find SDF associated with this compound
        matching = []

        for sdf in sdf_files:

            name = sdf.name.upper()

            if imp_id.upper() in name:
                matching.append(sdf)

        # If filename doesn't contain IMPHY ID,
        # search inside the compound's numeric folder
        if not matching:

            numeric_folder = OUTPUT_DIR / str(numeric_id)

            if numeric_folder.exists():
                matching = list(numeric_folder.rglob("*.sdf"))

        # copy to out_path

        if matching:

            sdf_file = matching[0]

            final_file = OUTPUT_DIR / f"{imp_id}.sdf"

            # Don't copy onto itself
            if sdf_file.resolve() != final_file.resolve():
                shutil.copy2(sdf_file, final_file)

            print(f"✓ SUCCESS: {final_file}")

            successful.append(imp_id)

        else:

            print(f"✗ SDF NOT FOUND: {imp_id}")

            failed.append(imp_id)

    except Exception as e:

        print(f"✗ ERROR: {imp_id}")
        print(e)

        failed.append(imp_id)


# final

print("\n")
print("=" * 60)
print("DOWNLOAD FINISHED")
print("=" * 60)

print(f"Total IDs : {len(ids)}")
print(f"Successful: {len(successful)}")
print(f"Failed    : {len(failed)}")

print(f"\nSDF files are in:")
print(OUTPUT_DIR)

# ============================================================
# SHOW FAILED IDs
# ============================================================

if failed:

    print("\nFailed IDs:")
    for imp_id in failed:
        print("  ", imp_id)

else:

    print("\n✓ All requested IDs were downloaded successfully!")
```
